### 2. ELIMINACIÓN DE REGISTROS QUE TENGAN NA EN LA COLUMNA "pm2.5"

[](La base de datos tomada para el taller corresponde a Beijing PM2.5 Analysis, y es un conjunto de datos que se realizó entre enero de 2010 y diciembre de 2014, momento en el cual la calidad del aire de Beijing era extremadamente mala. Por su parte, PM2.5 representa básicamente la calidad del aire cuando las personas apenas pueden ver los edificios, y parece un ambiente con neblina, pero en realidad no lo es, sino que es un caso de contaminación severa.)

[](EL conjunto de datos descargado tiene 43824 registros, cuyas variables son: "No.", "year", "month", "day", "hour", "pm2.5", "DEWP", "TEMP", "PRES", "cbdw", "Iws", "Is", "Ir")

[](A continuación se hace una descripción de las variables:)

[](No., corresponde al número de registro.)
[](year, corresponde al año del registro, expresado en número de cuatro dígitos.)
[](month, corresponde al mes del registro asociado al año, expresado como un número.)
[](day, corresponde al día del registro asociado a un mes y año, expresado como un número de acuerdo al calendario.)
[](hour, corresponde a la hora del registro asociado a un día, mes y año, expresado como un número entre 0 y 23.)
[](pm2.5, corresponde a la materia particulada 2.5 (por sus siglas en inglés: PARTICULATE MATTER) y son partículas muy pequeñas suspendidas en el aire que tienen un diámetro de menos de 2.5 micras, que incluye sustancias químicas orgánicas, polvo, hollín y metales, y puede provenir de todas las clases de combustión. Se considera un contaminante atmosférico por sus efectos nocivos sobre la salud.)
[](DEWP, corresponde al punto de rocío (tomado del inglés: DEW POINT), y se define como la temperatura atmosférica (que varía según la presión y la humedad) por debajo de la cual las gotas de agua comienzan a condensarse y se puede formar un rocío.)
[](TEMP, corresponde a la temperatura (del inglés TEMPERATURE))
[](PRES, corresponde a la presión (tomado del inglés: PRESSURE).)
[](cbdw, corresponde a la dirección combinada del viento (tomado del inglés: COMBINED WIND DIRECTION) y se expresa en función de la dirección combinada de los ejes principales cartesianos N-S y W-E, por ejemplo, NW.)
[](Iws, corresponde a la velocidad del viento acumulada (tomado del inglés: CUMULATED WIND SPEED) y se expresa en [m/s].)
[](Is, corresponde a las horas acumuladas de nieve (tomado del inglés: CUMULATE HOURS OF SNOW) y se expresa en horas.)
[](Ir, corresponde a las horas acumuladas de lluvia (tomada del inglés: CUMULATED HOURS OF RAIN) y se expresa en horas.)

[](En el conjunto de datos, para la variable "pm2.5", faltan muchos datos. En particular para la variable "pm.25", existiendo un faltante de 2076 datos)

[](Siendo la variable de estudio "pm2.5" se hace necesario eliminar los registros que tengan faltantes en "pm2.5", y que corresponden a los que tienen el valor de NA en dicho valor)

### ELIMINACIÓN DE REGISTROS FALTANTES EN "pm2.5"

[](Carga del Conjunto de Datos)

```python
##Se carga el conjunto de datos en la variable data
A = 'base_datos_Beijin_PM2.5.csv'
data = pd.read_csv(A, header=0)
print(data.shape)
data.head()
```

```python
##Se eliminan los registros "pm2.5" que son faltantes, es decir con valor NA
data_no_NA = data[pd.notnull(data['pm2.5'])]
print(data_no_NA.shape)
data_no_NA.head()
##Para crear un nuevo archivo extensión .csv, que no contiene registros NA, se puede activar la siguiente línea:
##data_no_NA.to_csv("data_no_NA.csv")
```

### 3. REALIZACIÓN DEL DESCRIBE DE TODAS LAS VARIABLES POR AÑO Y REALIZACIÓN DE UN GRÁFICO COMO BOXPLOT PARA LA COMPARACIÓN DE RESULTADOS DADOS

[](La función describe de Pandas regresa estadísticas descriptivas incluyendo: media, mediana, máximo, mínimo, desviación estándar y conteos para una columna en particular de los datos.)
[](Es importante mencionar que la función describe solo regresa los valores de estas estadísticas para las columnas numéricas.)

```python
data_no_NA.groupby('year').describe() ##Se calculan las estadísticas descriptivas del conjunto de datos
```

[](boxplot es un diagrama de caja y sirven para visualizar la medidad de qué tan bien distribuidos están los datos en un conjunto de datos, dividiendo el conjunto de datos en tres cuartiles, representando el mínimo, el máximo, la mediana, el primer cuartil y el tercer cuartil del conjunto de datos.)

```python
data_no_NA.boxplot(by ='year', column =['pm2.5'], grid = False)
```

### 4. AGRUPACIÓN POR AÑO DEL VALOR DE LA VARIABLE "pm2.5", Y CÁLCULO DEL PROMEDIO Y ORDENAMIENTO DE LA BASE DE DATOS DE MANERA ASCENDENTE POR EL PROMEDIO

```python
data_groupby_pm25 = data_no_NA.groupby(['year']).mean()
data_groupby_pm25.head()
##Para crear un nuevo archivo extensión .csv, que no contiene registros NA y que está agrupado por año del valor de la variable "pm2.5", y calculado su promedio, se puede activar la siguiente línea:
#data_groupby_pm25.to_csv("data_groupby_pm25.csv") ##
```

```python
data_groupby_pm25.sort_values(by=['pm2.5'],ascending=False) ##ordenando de manera ascendente con respecto a la variable "pm2.5"
```
