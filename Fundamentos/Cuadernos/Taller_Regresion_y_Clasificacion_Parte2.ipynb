
### 2.3. Elaboración de la Red Neuronal Artificial



Los modelos en Keras se definen como una secuencia de capas. Se creamos un modelo secuencial y se añaden capas una a una hasta que la topología de red satisface el resultado de la predicción (precisión). Lo primero que se hace es asegurar de que la capa de entrada tiene el número correcto de entradas. Esto se puede especificar al crear la primera capa con el argumento input_dim y establecerlo para las variables de entrada.

En este caso de estudio, se utiliza una estructura de red conectada con tres capas. Las capas completamente conectadas se densifican utilizando la clase Dense. Se puede especificar el número de neuronas en la capa como el primer argumento y especificar la función de activación usando el argumento de activación. Se usa la función de activación del rectificador (relu) en las dos primeras capas y la función de activación sigmoide en la capa de salida. Antes las funciones de activación sigmoide y tanh eran preferidas para todas las capas. Hoy en día, se observa un mejor rendimiento utilizando la función de activación del rectificador [[4]](https://unipython.com/desarrolla-primera-red-neural-python-keras-paso-paso/).


```python

##Fijación de las semillas aleatorias para la reproducibilidad:
np.random.seed(7)

##Creación del modelo:
model = Sequential()
model.add(Dense(20, input_dim=3, activation='relu'))
model.add(Dense(10, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```


Ahora que el modelo está definido, se puede compilar. La compilación del modelo utiliza las eficientes librerías numéricas de backend como Theano o TensorFlow. El backend elige automáticamente la mejor manera de representar la red para entrenar y hacer predicciones para que se ejecuten en su hardware, como CPU o GPU o incluso distribuidas.

Al compilar, se deben especificar algunas propiedades adicionales necesarias para la formación de la red. Entrenar una red significa encontrar el mejor conjunto de pesos para hacer predicciones para este problema. Se debe especificar la función de pérdida a utilizar para evaluar un conjunto de pesos, el optimizador utilizado para buscar a través de diferentes pesos para la red y cualquier métrica opcional que se quisiera recopilar y reportar durante el entrenamiento.

En este caso de estudio, se utilizará el error cuadrático medio, que es la función más comúnmente utilizada para un problema de  regresión y se define en Keras como “mse”. También utilizaremos como algoritmo de optimización el RMSprop que está diseñado para redes neuronales.

Finalmente, debido a que no es un problema de clasificación, los indicadores de evaluación utilizados en los problemas de regresión también son diferentes, por lo que el concepto de precisión no se aplica y en este caso se recopila y se reporta el error absoluto medio (mae) como la métrica.


```python

##Compilación del modelo:
model.compile(loss='mse', optimizer='rmsprop', metrics=['mae'])
```

```python

##Importar libreria para la separación de los datos de entrenamiento:
from sklearn.model_selection import train_test_split
##Utilizar 25% de los Datos para Entrenar los Modelos:
X_train, X_test, Y_train, Y_test = train_test_split(features, prices, test_size =0.25, random_state = 0)

print(X_train.shape)
print(X_train)

print(X_test.shape)
print(X_test)

print(Y_train.shape)
print(Y_train)

print(Y_test.shape)
print(Y_test)
```

```python

train_data = X_train
train_targets = Y_train
mean = train_data.mean(axis=0)
train_data -= mean
std = train_data.std(axis=0)
train_data /= std

test_data = X_test
test_data -= mean
test_data /= std

test_targets = Y_test

def smooth_curve (puntos, factor = 0.9):
        smoothed_points = []
        for point in points:
            if smoothed_points:
                previous = smoothed_points[-1]
                smoothed_points.append(previous * factor + point * (1 - factor))
            else:
                smoothed_points.append(point)
        return smoothed_points


model.fit(train_data,train_targets,epochs=80,batch_size=16,verbose=0)
test_mse_score,test_mae_score=model.evaluate(test_data,test_targets)
print(test_mae_score)
```
