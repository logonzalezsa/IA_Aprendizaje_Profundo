# DEFINICIÓN DE CPU, GPU & TPU

## INTRODUCCIÓN

[](En el aprendizaje de máquinas los algoritmos de entrenamiento requieren de elementos para dicho proceso. En estos términos se usa un hardwarde especial para tal fin, que son conocidos como CPU, GPU y TPU.)

## DIFERENCIACIÓN

[](En términos generales, las instrucciones se ejecutan y operan en una unidad central, que es denominada por sus iniciales en inglés como CPU).

[](Sin embargo, algunos procesos están relacionados, y en particular con el aprendizaje profundo, con tratamiento de imágenes y procesamiento de tensores, requeridos para dicho tratamiento de imágenes).

[](La unidad que acelera el procesamiento y tratamiento de imágenes es la GPU, por sus iniciales en inglés).

[](Por su parte, las operaciones del álgebra que requieren operaciones con tensores, se realiza en una unidad de procesamiento de tensores conocida como TPU)

## CONCLUSIONES

[](Se puede concluir que en un ámbito de disposición de hardwarde, Google provee de tres elementos poderosos para procesar los algoritmos requeridos en el aprendizaje de máquinas y aprendizaje profundo, desde una unidad central, la cual usa para el tratamiento de imágenes el GPU y para las operaciones con tensores el TPU)
